```python
def cleanname(nomprenom):
    """Nettoie le nom ou prénom : majuscules, sans accents, sans espaces ni tirets"""
    nomprenom = nomprenom.strip().replace(" ", "").replace("-", "")
    nomprenom = nomprenom.upper()

    
    remplacements = {
        "À": "A", "Â": "A", "Ä": "A", "Á": "A", "Ã": "A", "Å": "A",
        "É": "E", "È": "E", "Ê": "E", "Ë": "E",
        "Í": "I", "Î": "I", "Ï": "I", "Ì": "I",
        "Ô": "O", "Ö": "O", "Ò": "O", "Ó": "O", "Õ": "O",
        "Ù": "U", "Û": "U", "Ü": "U", "Ú": "U",
        "Ç": "C", "Ñ": "N"
    }

    for accent, normal in remplacements.items():
        nomprenom = nomprenom.replace(accent, normal)

    return nomprenom
```

```python
def extract_consonant(lastname):
    consonant = ""
    for letter in lastname:
        if letter in "BCDFGHJKLMNPQRSTVWXYZ":
            consonant += letter
    while len(consonant) < 3:
        consonant += "X"
    return consonant[:3]
```

```python
def extract_letter(firstname):
    firstname = cleanname(firstname)
    while len(firstname) < 3:
        firstname += "X"
    return firstname[:3]


gorde=extract_letter(firstname)
print(gorde)
```

```python
def generate_registration_number(nom, prenom, register):
    lastname = cleanname(nom)
    firstname = cleanname(prenom)

    for_lastname = extract_consonant(lastname)
    for_firstname = extract_letter(firstname)

    prefix = for_lastname + for_firstname

    if prefix not in register:
        register[prefix] = 0

    # Ensuite on incrémente toujours
    register[prefix] += 1
    
    number = str(register[prefix]).zfill(3)
    return prefix + number
```

```python
def Check_User_Input(fullname):
    while True:
        try:
            if fullname.replace(" ", "").replace("-", "").isalpha():
                #print(f" great: valid input '{user_input}'")
                return fullname
            else:
                raise ValueError(" invalid Input; please enter a valid name ")
        except ValueError as e:
            print(e)
            fullname = input("try again : ")
            
            
```

```python
def show_identical_prefixes(students):
    
    prefix_groups = {}

   
    for name, matricule in students.items():
        prefix = matricule[:-3]
        if prefix not in prefix_groups:
            prefix_groups[prefix] = []
        prefix_groups[prefix].append(name)

   
    print("\n=== Students with identical prefixes ===")
    found = False
    for prefix, names in prefix_groups.items():
        if len(names) > 1:
            found = True
            print(f"\nPrefix '{prefix}' is shared by:")
            for n in names:
                print(f"  ---> {n} ({students[n]})")

    if not found:
        print("No identical prefixes found.")
```

```python
if __name__ == "__main__":
    register = {}     # prifixes
    
    students = {}    # new dictionary with student details

    while True:
        lastname = input("enter the lastname of the student : ")
        lastnameS = Check_User_Input(lastname)
        firstname = input("enter the first name of the student : ")
        firstnameS = Check_User_Input(firstname)

        # Generation of the registration_number
        registration_number = generate_registration_number(lastnameS, firstnameS, register)
        print("\n The registration number is :", registration_number)

        # Enregistrement dans le dictionnair
        students[f"{lastnameS} {firstnameS}"] =registration_number

        
        other = input("\nDo you want to add another student  ? (y/n) : ").strip().lower()
        if other != "y":
            break
        print("")  

    # Print the final dictionnary 
    print("\n=== The full list of the students is ===")
    for identity, mat in students.items():
        print(f"{identity} ---> {mat}")

    show_identical_prefixes(students)
```
